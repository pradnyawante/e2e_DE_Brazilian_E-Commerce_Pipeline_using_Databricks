```python
sales = spark.table("workspace.ecommerce_schema.silver_sales")
```

```python
from pyspark.sql.functions import sum

gold_state_sales = sales.groupBy("customer_state") \
    .agg(sum("payment_value").alias("total_sales"))
```

```python
gold_state_sales.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.ecommerce_schema.gold_state_sales")
```
