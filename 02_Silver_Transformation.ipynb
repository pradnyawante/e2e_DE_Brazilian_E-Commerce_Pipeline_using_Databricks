```python
orders = spark.table("workspace.ecommerce_schema.bronze_orders")
order_items = spark.table("workspace.ecommerce_schema.bronze_order_items")
customers = spark.table("workspace.ecommerce_schema.bronze_customers")
products = spark.table("workspace.ecommerce_schema.bronze_products")
payments = spark.table("workspace.ecommerce_schema.bronze_payments")
```

```python
from pyspark.sql.functions import col

silver_orders = orders.filter(col("order_status") == "delivered")
```

```python
silver_orders_customers = silver_orders.join(
    customers,
    "customer_id",
    "inner"
)
```

```python
silver_full = silver_orders_customers.join(
    payments,
    "order_id",
    "inner"
)
```

```python
silver_full.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.ecommerce_schema.silver_sales")
```
